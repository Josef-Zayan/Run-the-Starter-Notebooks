## 3. Quick look at the data (2-3 real numbers)

Of the 30,000 pages in the starter dataset, 16,262 (54.2%) are currently
declining — too many for an editor to review one by one. A simple
"stale and visible" rule (180+ days since update, 500+ impressions) is
very precise — 16 of the 17 pages it flags are declining — but it only
flags 17 pages, because just 174 pages are stale at all. The rule is
sharp but narrow; the question for the next 7 weeks is whether a ranked
model can find more of the declining pages without losing that precision.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = len(df)
is_declining = df["trend_direction"].str.lower() == "down"
is_stale = df["days_since_last_update"] >= 180
is_visible = df["impressions_90d"] >= 500
rule_mask = is_stale & is_visible

declining = is_declining.sum()
stale = is_stale.sum()
stale_and_visible = rule_mask.sum()
declining_in_rule = (rule_mask & is_declining).sum()

print(f"Total pages in dataset: {total_pages}")
print(f"Pages currently declining: {declining} ({declining/total_pages*100:.1f}%)")
print(f"Pages stale (180+ days) at all: {stale}")
print(f"Pages that are both stale and visible (500+ impressions): {stale_and_visible}")
print(f"Of those {stale_and_visible}, declining: {declining_in_rule}")
```

## 4. Careful words: what I can and can't claim

What this work CAN claim:
- Observed patterns in the starter data (e.g. "declining pages in this sample tend to...").
- Directional signals: which page features are associated with decline.
- Decision support: a ranked list that helps an editor choose where to look first.

What this work will NEVER claim:
- Causal proof: I can't say a page declined *because of* its age or word count — this is observational data, not an experiment.
- "Predicting Google's algorithm": I only see outcomes, not how Google ranks pages.
- That refreshing a flagged page will bring its traffic back — a high score means "worth reviewing", not "guaranteed to recover".

One honest limit to keep in mind: the "declining" label comes from trend_direction in the same 90-day window, so it describes pages that are already declining, not a prediction of future decline. The results also come from a 30,000-page sample of 32 clients, so they may not hold for every site.
